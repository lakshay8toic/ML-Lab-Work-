## pandas: Series and DataFrame

`DataFrame` and `Series` are the two data structures that pandas provides for processing data.

### 1) Series

A one-dimensional array capable of storing a variety of data types. The term **index** refers to the row labels of a Series. Multiple columns cannot be included in a Series.

Only one parameter is required — **data**: it can be any list, dictionary, or scalar value.

```python
import pandas as pd
```

### Creating Series from Array

It has the following parameters:

- **data:** It can be any list, dictionary, or scalar value.
- **index:** The value of the index should be unique and hashable. It must be of the same length as data. If we do not pass any index, the default `np.arange(n)` will be used.
- **dtype:** It refers to the data type of the series.
- **copy:** It is used for copying the data.

```python
import pandas as pd
import numpy as np

info = np.array(['S', 'I', 'V', 'A'])

a = pd.Series(info)

print(a)
```

### DataFrame

A pandas DataFrame is a widely used data structure which works with a two-dimensional array with labelled axes (rows and columns). A DataFrame is defined as a standard way to store data that has two different indexes, i.e. **row index** and **column index**.

- The columns can be heterogeneous types like int, bool, and so on.
- It can be seen as a dictionary of Series structures where both the rows and columns are indexed. It is denoted as "columns" in case of columns and "index" in case of rows.

**Parameters & Description:**

- **data:** It consists of different forms like ndarray, series, map, constants, lists, array.
- **index:** The default `np.arange(n)` index is used for the row labels if no index is passed.
- **columns:** The default syntax is `np.arange(n)` for the column labels. It shows only true if no index is passed.
- **dtype:** It refers to the data type of each column.
- **copy():** It is used for copying the data.

We can create a DataFrame in the following ways: `dict`, lists, NumPy ndarrays, and Series.

#### Create an empty DataFrame

```python
import pandas as pd

df = pd.DataFrame()

print(df)
```

#### Create a DataFrame using a List

```python
import pandas as pd

# a list of strings
x = ['Siva', 'Rama', 'Krishna']

# Calling DataFrame constructor on list
df = pd.DataFrame(x)

print(df)
```

#### Create a DataFrame from a Dict of ndarrays / Lists

```python
import pandas as pd

info = {'ID': [1001, 1002, 1003], 'Department': ['B.Tech', 'M.Tech', 'PhD']}
df = pd.DataFrame(info)
print(df)
```

#### Create a DataFrame from a Dict of Series

```python
import pandas as pd

info = {'one': pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f']),
        'two': pd.Series([1, 2, 3, 4, 5, 6, 7, 8], index=['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h'])}
d1 = pd.DataFrame(info)
print(d1)
```

### Column Selection

```python
import pandas as pd

info = {'one': pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f']),
        'two': pd.Series([1, 2, 3, 4, 5, 6, 7, 8], index=['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h'])}
d1 = pd.DataFrame(info)
print(d1['one'])
print(d1['two'])
```

### Column Addition

```python
import pandas as pd

info = {'one': pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
        'two': pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print(df)

# Add a new column to an existing DataFrame object
print("Add new column by passing series")
df['three'] = pd.Series([20, 40, 60], index=['a', 'b', 'c'])
print(df)

print("Add new column using existing DataFrame columns")
df['four'] = df['one'] + df['three']
print(df)
```

### Column Deletion

```python
import pandas as pd

info = {'one': pd.Series([1, 2], index=['a', 'b']),
        'two': pd.Series([1, 2, 3], index=['a', 'b', 'c'])}

df = pd.DataFrame(info)
print("The DataFrame:")
print(df)

# using del function
print("Delete the first column:")
del df['one']
print(df)

# using pop function
print("Delete the another column:")
df.pop('two')
print(df)
```

### Selection by Label

We can select any row by passing the row label to a `loc` function.

```python
import pandas as pd

info = {'one': pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
        'two': pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}

df = pd.DataFrame(info)
print("Data frame:")
print(df)
print("b index elements are:")
print(df.loc['b'])
print("e index elements are:")
print(df.loc['e'])
```

The rows can also be selected by passing the integer location to an `iloc` function.

```python
import pandas as pd

info = {'one': pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
        'two': pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print(df)
print(df.iloc[3])
```

### Slice Rows

It is another method to select multiple rows using the `:` operator.

```python
import pandas as pd

info = {'one': pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
        'two': pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print(df)
print(df[2:5])
```

### Addition of rows

We can add new rows to a DataFrame. (`DataFrame.append()` was removed in pandas 2.0, so `pd.concat` is used here — it adds the new rows at the end.)

```python
import pandas as pd

# Assuming d and d2 are defined DataFrames
d = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
d2 = pd.DataFrame({'A': [7, 8, 9], 'B': [10, 11, 12]})

# Replace append with concat
d = pd.concat([d, d2])

print(d)
```

### Deletion of rows

We can delete or drop any row from a DataFrame using the index label. If the label is duplicated, then multiple rows will be deleted.

```python
import pandas as pd

a_info = pd.DataFrame([[4, 5], [6, 7]], columns=['x', 'y'])
b_info = pd.DataFrame([[8, 9], [10, 11]], columns=['x', 'y'])
a_info = pd.concat([a_info, b_info])
print(a_info)

# Drop rows with label 0
a_info = a_info.drop(0)
print("Drop the row with label 0")
print(a_info)
```

### Selecting a column and a single element

```python
import pandas as pd

data = {'Gender': ['F', 'M', 'M'], 'Emp_ID': ['E01', 'E02', 'E03'], 'Age': [25, 27, 25]}

# We want to order the columns, so let's specify them in the columns parameter
df = pd.DataFrame(data, columns=['Emp_ID', 'Gender', 'Age'])
print(df)            # output all data
print(df.Emp_ID)     # output all Emp_ID (entire column)
print(df.Emp_ID[0])  # output first Emp_ID
```

### Working with a CSV File

**Run the setup cell below once** — it fetches the same Iris dataset and saves it as

the cells that follow works unchanged.

*(If you have your own `iris.csv`, upload it with the Files panel on the left of
Colab instead and skip this cell.)*

```python
# --- SETUP: make iris.csv available in this Colab session ---
import pandas as pd

URL = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

try:
    iris = pd.read_csv(URL)
except Exception as err:
    # No internet? Fall back to the copy bundled with scikit-learn.
    print("Download failed (", type(err).__name__, ") - using the scikit-learn copy instead.")
    from sklearn.datasets import load_iris
    raw = load_iris(as_frame=True)
    iris = raw.frame.rename(columns={
        'sepal length (cm)': 'sepal_length', 'sepal width (cm)': 'sepal_width',
        'petal length (cm)': 'petal_length', 'petal width (cm)': 'petal_width'})
    iris['species'] = raw.target.map(dict(enumerate(raw.target_names)))
    iris = iris.drop(columns=['target'])

iris.to_csv('iris.csv', index=False)
print("iris.csv is ready -", iris.shape[0], "rows,", iris.shape[1], "columns")
iris.head()
```

### DataFrame.drop_duplicates()

```python
import pandas as pd

emp = {"Name": ["Siva", "Rama", "Krishna", "Siva"],
       "Age": [21, 32, 29, 21]}
info = pd.DataFrame(emp)
print(info)

info = info.drop_duplicates()
print("After removing the duplicates the dataframe is :")
print(info)
```

### Null value identification in a given dataset

```python
import pandas as pd
import numpy as np

info = pd.DataFrame({"Person": ["Siva", "Pavan", "Gopal", "Raghu"],
                     "Age": [27., 29, np.nan, 32]})
print(info)
print(info.count())
print("null values")
print(info.isnull())
print(info.isnull().sum())
```

### Sorting order

```python
import pandas as pd
import numpy as np

info = pd.DataFrame({"Person": ["Siva", "Pavan", "Gopal", "Raghu"],
                     "Age": [27., 29, np.nan, 32]})
print(info)
print("Sorting Order")
print(info.sort_index())
print("Descending Order")
print(info.sort_index(ascending=False))
```

### Aggregation Functions

Uses the `iris.csv` created by the setup cell above.

```python
import pandas as pd

df = pd.read_csv('iris.csv')
print(df)

mean_df = df.groupby('species').mean()
print("Mean values for each species:")
print(mean_df)

sum_df = df.groupby('species').sum()
print("\nSum of values for each species:")
print(sum_df)
```

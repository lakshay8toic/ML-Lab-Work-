```python
import pandas as pd
import numpy as np

data = pd.read_csv('final_vg.csv')
data.head()
```

```python
cat_count= data['Genre'].value_counts()
```

```python
cat_count
```

```python
cat_count.index
```

```python
x_bar= cat_count.index
y_bar= cat_count
plt.bar(x_bar, y_bar)
```

```python
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar)
```

```python
#rotate the x labels
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar)
plt.xticks(rotation=90, fontsize=12)
```

```python
#Change the width of these bars
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar, width=0.2)
plt.xticks(rotation=90, fontsize=12)
```

```python
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar, width=0.2, color='orange')
plt.xticks(rotation=90, fontsize=12)
plt.title("Games per Genre", fontsize= 15)
plt.xlabel("Genre")
plt.ylabel("Count of Games")
```

```python
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar, width=0.2, color='orange')
plt.xticks(rotation=90, fontsize=12)
plt.title("Games per Genre", fontsize= 15)
plt.xlabel("Genre")
plt.ylabel("Count of Games")
plt.show()
```
